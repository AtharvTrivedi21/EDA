```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

```python
# Load the dataset
df = pd.read_csv('WHR2024.csv')

# Inspect the dataset
df.head()
```

```python
# Simple Linear Regression: Using one explanatory variable (Log GDP per capita)
X_simple = df['Explained by: Log GDP per capita']
y = df['Ladder score']

# Reshape X_simple for statsmodels
X_simple = sm.add_constant(X_simple)  # Adds a constant term to the model
```

```python
# Multiple Linear Regression: Using multiple explanatory variables
X_multiple = df[['Explained by: Log GDP per capita',
                 'Explained by: Social support',
                 'Explained by: Healthy life expectancy',
                 'Explained by: Freedom to make life choices',
                 'Explained by: Generosity',
                 'Explained by: Perceptions of corruption']]

# Add constant to the model for intercept
X_multiple = sm.add_constant(X_multiple)
```

```python
# Check for NaN and Infinite values in the independent variables (X) and dependent variable (y)
print("Missing values in independent variables:")
print(X_simple.isnull().sum())

print("Missing values in dependent variable:")
print(y.isnull().sum())

print("Checking for infinite values in independent variables:")
print(np.isinf(X_simple).sum())

print("Checking for infinite values in dependent variable:")
print(np.isinf(y).sum())
```

```python
# Drop rows with missing values
df_cleaned = df.dropna()

# Re-define X and y after dropping missing values
X_simple = df_cleaned['Explained by: Log GDP per capita']
y = df_cleaned['Ladder score']

# Reshape for statsmodels
X_simple = sm.add_constant(X_simple)
```

```python
# Fill missing values with the median (for example)
df['Explained by: Log GDP per capita'].fillna(df['Explained by: Log GDP per capita'].median(), inplace=True)
df['Ladder score'].fillna(df['Ladder score'].median(), inplace=True)
```

```python
# Check data types of all columns
print(df.dtypes)
```

```python
# Select independent (X) and dependent (y) variables for Simple Linear Regression
X_simple = df['Explained by: Log GDP per capita']
y = df['Ladder score']

# Drop rows with missing values
df_cleaned = df.dropna(subset=['Explained by: Log GDP per capita', 'Ladder score'])

X_simple = df_cleaned['Explained by: Log GDP per capita']
y = df_cleaned['Ladder score']

# Add constant term for statsmodels
X_simple = sm.add_constant(X_simple)

# Fit the Simple Linear Regression model
model_simple = sm.OLS(y, X_simple).fit()

# Display the summary of the model
print(model_simple.summary())
```

```python
# Select independent (X) and dependent (y) variables for Multiple Linear Regression
X_multiple = df[['Explained by: Log GDP per capita',
                 'Explained by: Social support',
                 'Explained by: Healthy life expectancy',
                 'Explained by: Freedom to make life choices',
                 'Explained by: Generosity',
                 'Explained by: Perceptions of corruption']]

# Drop rows with missing values
df_cleaned = df.dropna(subset=X_multiple.columns.tolist() + ['Ladder score'])

X_multiple = df_cleaned[['Explained by: Log GDP per capita',
                         'Explained by: Social support',
                         'Explained by: Healthy life expectancy',
                         'Explained by: Freedom to make life choices',
                         'Explained by: Generosity',
                         'Explained by: Perceptions of corruption']]

y = df_cleaned['Ladder score']

# Add constant term for statsmodels
X_multiple = sm.add_constant(X_multiple)

# Fit the Multiple Linear Regression model
model_multiple = sm.OLS(y, X_multiple).fit()

# Display the summary of the model
print(model_multiple.summary())
```

```python
# Predict using Simple Linear Regression model
# Predict the Ladder score for the existing data
y_pred_simple = model_simple.predict(X_simple)

# Print the first few predictions
print("\nFirst few predictions from Simple Linear Regression:")
print(y_pred_simple.head())
```

```python
# Predict using Multiple Linear Regression model
# Predict the Ladder score for the existing data
y_pred_multiple = model_multiple.predict(X_multiple)

# Print the first few predictions
print("\nFirst few predictions from Multiple Linear Regression:")
print(y_pred_multiple.head())

```

```python
# Drop rows with missing values for both independent and dependent variables
df_cleaned = df.dropna(subset=['Ladder score',
                               'Explained by: Log GDP per capita',
                               'Explained by: Social support',
                               'Explained by: Healthy life expectancy',
                               'Explained by: Freedom to make life choices',
                               'Explained by: Generosity',
                               'Explained by: Perceptions of corruption'])

# Re-define X and y after cleaning the data
X_simple = df_cleaned['Explained by: Log GDP per capita']
y = df_cleaned['Ladder score']

X_multiple = df_cleaned[['Explained by: Log GDP per capita',
                         'Explained by: Social support',
                         'Explained by: Healthy life expectancy',
                         'Explained by: Freedom to make life choices',
                         'Explained by: Generosity',
                         'Explained by: Perceptions of corruption']]

# Add constant term for statsmodels
X_simple = sm.add_constant(X_simple)
X_multiple = sm.add_constant(X_multiple)
```

```python
# Fit Simple Linear Regression model
model_simple = sm.OLS(y, X_simple).fit()

# Fit Multiple Linear Regression model
model_multiple = sm.OLS(y, X_multiple).fit()
```

```python
# Predict using Simple Linear Regression model
y_pred_simple = model_simple.predict(X_simple)

# Predict using Multiple Linear Regression model
y_pred_multiple = model_multiple.predict(X_multiple)

# Print the first few predictions from both models
print("\nFirst few predictions from Simple Linear Regression:")
print(y_pred_simple.head())

print("\nFirst few predictions from Multiple Linear Regression:")
print(y_pred_multiple.head())
```

```python
# Evaluate the Simple Linear Regression model
mae_simple = mean_absolute_error(y, y_pred_simple)
mse_simple = mean_squared_error(y, y_pred_simple)
rmse_simple = np.sqrt(mse_simple)
r2_simple = r2_score(y, y_pred_simple)

print(f"\nSimple Linear Regression Evaluation Metrics:")
print(f"Mean Absolute Error (MAE): {mae_simple:.4f}")
print(f"Mean Squared Error (MSE): {mse_simple:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse_simple:.4f}")
print(f"R-squared: {r2_simple:.4f}")

# Evaluate the Multiple Linear Regression model
mae_multiple = mean_absolute_error(y, y_pred_multiple)
mse_multiple = mean_squared_error(y, y_pred_multiple)
rmse_multiple = np.sqrt(mse_multiple)
r2_multiple = r2_score(y, y_pred_multiple)

print(f"\nMultiple Linear Regression Evaluation Metrics:")
print(f"Mean Absolute Error (MAE): {mae_multiple:.4f}")
print(f"Mean Squared Error (MSE): {mse_multiple:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse_multiple:.4f}")
print(f"R-squared: {r2_multiple:.4f}")
```

# Conclusion Report
## Overview:
- We used a dataset containing variables related to the World Happiness Report 2024 to implement Simple Linear Regression and Multiple Linear Regression models to predict the Ladder Score (a measure of subjective well-being in each country) based on various independent factors. We performed the following steps:

- Data Cleaning: We removed rows with missing values for the relevant columns to ensure a consistent dataset for modeling.
- Modeling: We implemented both Simple and Multiple Linear Regression models to predict the Ladder Score.
- Evaluation: We evaluated the models using standard metrics such as Mean Absolute Error (MAE), Mean Squared Error (MSE), Root Mean Squared Error (RMSE), and R-squared.

## Key Findings:
### Simple Linear Regression:
- We used Explained by: Log GDP per capita as the sole independent variable to predict the Ladder Score.
- The model's R-squared value indicated that the independent variable explains a significant portion of the variation in the dependent variable. However, it was less than perfect, suggesting other factors influence the Ladder Score.

### Multiple Linear Regression:

- We included multiple independent variables such as Social Support, Healthy Life Expectancy, Freedom to Make Life Choices, Generosity, and Perceptions of Corruption.
- The R-squared value improved significantly compared to the Simple Linear Regression model, indicating that the model explains more of the variance in Ladder Score when multiple factors are considered.
- This model provides a more holistic view of the factors contributing to happiness, as it accounts for more than just Log GDP per capita.

### Model Evaluation Metrics:
- R-squared: Both models demonstrated a reasonable ability to explain the variance in Ladder Score, with the Multiple Linear Regression model showing a higher value, indicating better performance.
- MAE, MSE, RMSE: The error metrics (MAE, MSE, RMSE) were relatively low for both models, with the Multiple Linear Regression model outperforming the Simple Linear Regression model. This suggests that using multiple predictors improves the model's accuracy.
- p-values: The p-values for the independent variables in the Multiple Linear Regression model were significant, indicating that each predictor has a meaningful relationship with the Ladder Score.

### Model Insights:
- The Log GDP per capita is an important predictor in explaining the Ladder Score, but it doesn't capture all the factors that affect happiness. The Multiple Linear Regression model, which includes additional predictors, is more comprehensive and provides a better fit to the data.
- Social Support, Healthy Life Expectancy, and Freedom to Make Life Choices were all significant factors in predicting the Ladder Score, highlighting the importance of social and health factors in subjective well-being.

### Conclusion:
- The analysis demonstrates that both Simple Linear Regression and Multiple Linear Regression are valuable tools for predicting the Ladder Score. However, the Multiple Linear Regression model offers a more accurate and nuanced prediction by incorporating multiple factors that influence happiness beyond just economic measures like GDP.

- This study emphasizes the importance of considering a variety of factors when evaluating subjective well-being and suggests that countries aiming to improve happiness levels should focus not only on economic growth but also on improving social support systems, healthcare, and personal freedoms.
