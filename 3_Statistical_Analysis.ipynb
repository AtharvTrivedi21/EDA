```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
#generate 500 random values between 1 and 10000
data = np.random.uniform(1, 10000, 500)

#generated numbers
print("Generated 500 random values:")
print(data)
```

```python
#compute statistics
mean_val = np.mean(data)
variance_val = np.var(data, ddof=0)  # Population variance
std_dev_val = np.std(data, ddof=0)   # Population standard deviation

#compute z-scores
z_scores = (data - mean_val) / std_dev_val
```

```python
#display statistics
print(f"Mean: {mean_val}")
print(f"Variance: {variance_val}")
print(f"Standard Deviation: {std_dev_val}")
```

```python
#display z-scores
print("Z-Scores:")
print(z_scores)
```

```python
#read number of bins from user
n_bins = int(input("Enter the number of bins for the histogram: "))
```

```python
# Plot Scatter plot
plt.figure(figsize=(18, 6))  # Increase figure width for better visibility

plt.subplot(1, 3, 1)
plt.scatter(range(len(data)), data, color='blue', alpha=0.6)
plt.axhline(mean_val, color='red', linestyle='dashed', label='Mean')
plt.xlabel("Index")
plt.ylabel("Value")
plt.title("Scatter Plot")
plt.legend()

# Plot Histogram
plt.subplot(1, 3, 2)
plt.hist(data, bins=n_bins, color='skyblue', edgecolor='black', alpha=0.7)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram")

# Plot Gaussian Density Function
plt.subplot(1, 3, 3)
x = np.linspace(min(data), max(data), 1000)
pdf = stats.norm.pdf(x, mean_val, std_dev_val)
plt.plot(x, pdf, color='green', linewidth=2, label='Gaussian PDF')
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Density Plot (Gaussian PDF)")
plt.legend()

plt.tight_layout()
plt.show()
```
